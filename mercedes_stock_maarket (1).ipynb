```python
daily_volatility = df['Daily_Return'].std()

print("Daily Volatility:", daily_volatility)
```

```python
annual_volatility = (
    df['Daily_Return'].std() * (252 ** 0.5)
)

print("Annualized Volatility:", annual_volatility)
```

```python
wealth = (
    1 + df['Daily_Return']
).cumprod()
```

```python
running_peak = wealth.cummax()
```

```python
drawdown = (
    wealth - running_peak
) / running_peak
```

```python
max_drawdown = drawdown.min()

print("Maximum Drawdown:", max_drawdown)
```

```python
print(
    "Maximum Drawdown (%):",
    max_drawdown * 100
)
```

```python
plt.figure(figsize=(14,6))

plt.plot(
    df['Date'],
    drawdown * 100
)

plt.title('Mercedes-Benz Historical Drawdown')
plt.xlabel('Date')
plt.ylabel('Drawdown (%)')

plt.show()
```

```python
import yfinance as yf
```

```python
stocks = [
    "MBG.DE",
    "BMW.DE",
    "VOW3.DE",
    "7203.T",
    "TSLA"
]
```

```python
portfolio_data = yf.download(
    stocks,
    start="2020-01-01",
    end="2026-02-26",
    auto_adjust=True
)
```

```python
portfolio_data.head()
```

```python
portfolio_data.shape
```

```python
portfolio_prices = portfolio_data["Close"]
```

```python
portfolio_prices.head()
```

```python
portfolio_prices.isnull().sum()
```

```python
portfolio_prices = portfolio_prices.dropna()
```

```python
portfolio_prices.isnull().sum()
```

```python
portfolio_prices.shape
```

```python
print("Start:", portfolio_prices.index.min())
print("End:", portfolio_prices.index.max())
```

```python
portfolio_returns = portfolio_prices.pct_change()
```

```python
portfolio_returns = portfolio_returns.dropna()
```

```python
portfolio_returns.head()
```

```python
correlation_matrix = portfolio_returns.corr()
```

```python
correlation_matrix
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(9,7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Stock Return Correlation Matrix")

plt.show()
```

```python
weights = {
    "MBG.DE": 0.20,
    "BMW.DE": 0.20,
    "VOW3.DE": 0.20,
    "7203.T": 0.20,
    "TSLA": 0.20
}
```

```python
sum(weights.values())
```

```python
portfolio_daily_return = sum(
    portfolio_returns[stock] * weight
    for stock, weight in weights.items()
)
```

```python
portfolio_daily_return.head()
```

```python
portfolio_growth = (
    1 + portfolio_daily_return
).cumprod()
```

```python
portfolio_growth.head()
```

```python
plt.figure(figsize=(14,6))

plt.plot(
    portfolio_growth.index,
    portfolio_growth,
    label="Hypothetical Portfolio"
)

plt.title("Hypothetical Portfolio Growth")
plt.xlabel("Date")
plt.ylabel("Growth of ₹1")

plt.legend()
plt.show()
```

```python
portfolio_annual_return = (
    portfolio_daily_return.mean() * 252
)
```

```python
print(
    "Annualized Portfolio Return:",
    round(portfolio_annual_return * 100, 2),
    "%"
)
```

```python
portfolio_annual_volatility = (
    portfolio_daily_return.std() * (252 ** 0.5)
)
```

```python
portfolio_peak = portfolio_growth.cummax()
```

```python
portfolio_drawdown = (
    portfolio_growth - portfolio_peak
) / portfolio_peak
```

```python
portfolio_max_drawdown = portfolio_drawdown.min()
```

```python
print(
    "Maximum Portfolio Drawdown:",
    round(portfolio_max_drawdown * 100, 2),
    "%"
)
```

```python
plt.figure(figsize=(14,6))

plt.plot(
    portfolio_drawdown.index,
    portfolio_drawdown * 100
)

plt.title("Portfolio Historical Drawdown")
plt.xlabel("Date")
plt.ylabel("Drawdown (%)")

plt.show()
```

```python
portfolio_summary = pd.DataFrame({
    "Metric": [
        "Annualized Return",
        "Annualized Volatility",
        "Maximum Drawdown"
    ],
    "Value": [
        portfolio_annual_return * 100,
        portfolio_annual_volatility * 100,
        portfolio_max_drawdown * 100
    ]
})

portfolio_summary
```

```python
individual_annual_return = portfolio_returns.mean() * 252 * 100

individual_annual_return
```

```python
individual_annual_volatility = (
    portfolio_returns.std() * (252 ** 0.5) * 100
)

individual_annual_volatility
```

```python
individual_drawdowns = {}

for stock in stocks:
    stock_growth = (
        1 + portfolio_returns[stock]
    ).cumprod()

    stock_peak = stock_growth.cummax()

    stock_drawdown = (
        stock_growth - stock_peak
    ) / stock_peak

    individual_drawdowns[stock] = stock_drawdown.min() * 100
```

```python
individual_drawdowns
```

```python
stock_comparison = pd.DataFrame({
    "Annualized Return (%)": individual_annual_return,
    "Annualized Volatility (%)": individual_annual_volatility,
    "Maximum Drawdown (%)": pd.Series(individual_drawdowns)
})

stock_comparison
```

```python
portfolio_row = pd.DataFrame({
    "Annualized Return (%)": [portfolio_annual_return * 100],
    "Annualized Volatility (%)": [portfolio_annual_volatility * 100],
    "Maximum Drawdown (%)": [portfolio_max_drawdown * 100]
}, index=["Portfolio"])
```

```python
final_comparison = pd.concat([
    stock_comparison,
    portfolio_row
])
```

```python
final_comparison
```

```python
plt.figure(figsize=(10,7))

plt.scatter(
    individual_annual_volatility,
    individual_annual_return
)

for stock in stocks:
    plt.annotate(
        stock,
        (
            individual_annual_volatility[stock],
            individual_annual_return[stock]
        )
    )

plt.xlabel("Annualized Volatility (%)")
plt.ylabel("Annualized Return (%)")
plt.title("Historical Risk vs Return")

plt.show()
```

```python
plt.figure(figsize=(15,7))

for stock in stocks:
    stock_growth = (
        1 + portfolio_returns[stock]
    ).cumprod()

    plt.plot(
        stock_growth,
        label=stock
    )

plt.plot(
    portfolio_growth,
    label="Portfolio",
    linewidth=3
)

plt.title("Individual Stocks vs Hypothetical Portfolio")
plt.xlabel("Date")
plt.ylabel("Growth of ₹1")

plt.legend()
plt.show()
```
